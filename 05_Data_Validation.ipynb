# Data Validation

## 1. What Is Data Validation?

Data validation is the process of checking whether data follows predefined rules, conditions, and requirements.

It helps us identify invalid, inconsistent, incorrect, or unexpected values before using the data for analysis or Machine Learning.

**Example:**

Suppose a dataset contains employee information.

| Employee | Age | Salary | Department |
| -------- | --: | -----: | ---------- |
| Arun     |  25 |  30000 | IT         |
| Priya    |  -5 |  40000 | HR         |
| Kumar    |  30 |  -1000 | Finance    |
| Divya    |  28 |  35000 | Unknown    |

Here:

* Age `-5` is invalid because age cannot be negative.
* Salary `-1000` is invalid because salary cannot be negative.
* Department `Unknown` may be invalid if it is not an approved department.

Data validation helps identify these problems.

## 2. Why Is Data Validation Important?

Data validation is important because:

* It helps identify invalid data.
* It improves data quality and consistency.
* It prevents incorrect values from entering the analysis.
* It reduces errors during model training.
* It helps maintain reliable Machine Learning predictions.
* It ensures that data follows business requirements.

**AI/ML example:**

Suppose we are developing a model to predict employee salaries. If the dataset contains negative salaries or invalid experience values, the model may learn incorrect patterns.

Validating the data helps us identify these problems before training the model.

## 3. Types of Data Validation

### A. Range Validation

Checks whether a numerical value falls within an acceptable range.

Examples:

* Age must be greater than or equal to 0.
* Percentage must be between 0 and 100.
* Product quantity must not be negative.

### B. Data Type Validation

Checks whether a column contains the expected data type.

Examples:

* Age should be numeric.
* Joining date should be a date.
* Employee name should be text.

### C. Format Validation

Checks whether a value follows the required format.

Examples:

* Email addresses should follow an accepted email format.
* Phone numbers should follow the expected structure.
* Dates should use an accepted date format.

A format check does not necessarily prove that an email address or phone number actually exists.

# Demonstration the Concept – Data Validation

## Real-World Example: Employee Data Validation

Imagine a company maintains an employee dataset containing the following information:

* Employee ID
* Employee Name
* Age
* Salary
* Department
* Joining Date

Before using this dataset for employee analytics or Machine Learning, the company must verify that the information follows its requirements.

### Sample Dataset

| Employee ID | Employee Name |     Age | Salary | Department | Joining Date |
| ----------: | ------------- | ------: | -----: | ---------- | ------------ |
|         101 | Arun          |      25 |  30000 | IT         | 2023-01-10   |
|         102 | Priya         |      -5 |  40000 | HR         | 2022-05-15   |
|         103 | Kumar         |      30 |  -1000 | Finance    | 2021-08-20   |
|         104 | Divya         |      28 |  35000 | Unknown    | 2023-11-05   |
|         105 | Ravi          | Missing |  45000 | IT         | 2024-02-10   |
|         101 | Meena         |      32 |  50000 | Sales      | 2024-04-15   |

### Problems in the Dataset

**Problem 1: Invalid Age**

Priya's age is `-5`.

Rule: Age must be greater than or equal to zero.

**Problem 2: Invalid Salary**

Kumar's salary is `-1000`.

Rule: Salary must be greater than or equal to zero.

**Problem 3: Invalid Category**

Divya's department is `Unknown`.

Rule: Department must belong to the approved categories: IT, HR, Finance, or Sales.

**Problem 4: Missing Value**

Ravi's age is missing.

Rule: Employee age is required for this example.

**Problem 5: Duplicate Employee ID**

Employee ID `101` appears twice.

Rule: Employee ID must be unique.

However, we must investigate the duplicate before deleting anything. It could be a duplicate record, or it could indicate an incorrect employee ID.

## Validation Rules

For this example, we will define these rules:

1. Employee ID must not be missing.
2. Employee ID must be unique.
3. Employee Name must not be missing.
4. Age must be present and non-negative.
5. Salary must be present and non-negative.
6. Department must be one of the approved categories.
7. Joining Date must contain a valid date.

These are example business rules. Real organizations may require additional rules, such as a reasonable age range or salary limits.

## Expected Result

After validation, we should know:

* Which records have invalid ages.
* Which records have invalid salaries.
* Which records contain unapproved departments.
* Which records have missing required values.
* Which employee IDs are duplicated.
* Whether the joining dates are valid.

**Important:** Validation does not automatically mean deleting every record that fails a rule. We must investigate the issue and select an appropriate action.


```python
import pandas as pd

# Step 1: Create a sample employee dataset
data = {
    "Employee_ID": [101, 102, 103, 104, 105, 101],
    "Employee_Name": ["Arun", "Priya", "Kumar", "Divya", "Ravi", "Meena"],
    "Age": [25, -5, 30, 28, None, 32],
    "Salary": [30000, 40000, -1000, 35000, 45000, 50000],
    "Department": ["IT", "HR", "Finance", "Unknown", "IT", "Sales"],
    "Joining_Date": [
        "2023-01-10",
        "2022-05-15",
        "2021-08-20",
        "2023-11-05",
        "2024-02-10",
        "invalid_date"
    ]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Step 2: Define validation rules
allowed_departments = ["IT", "HR", "Finance", "Sales"]

# Step 3: Validate Employee ID
invalid_employee_id = df["Employee_ID"].isnull()

# Identify all rows containing a repeated Employee ID
duplicate_employee_id = df["Employee_ID"].duplicated(keep=False)

# Step 4: Validate Employee Name
invalid_employee_name = (
    df["Employee_Name"].isnull()
    | df["Employee_Name"].astype("string").str.strip().eq("")
)

# Step 5: Validate Age
invalid_age = df["Age"].isnull() | (df["Age"] < 0)

# Step 6: Validate Salary
invalid_salary = df["Salary"].isnull() | (df["Salary"] < 0)

# Step 7: Validate Department
invalid_department = (
    df["Department"].isnull()
    | ~df["Department"].isin(allowed_departments)
)

# Step 8: Validate Joining Date
df["Joining_Date"] = pd.to_datetime(
    df["Joining_Date"],
    errors="coerce"
)

invalid_joining_date = df["Joining_Date"].isnull()

# Step 9: Create a validation report
validation_report = pd.DataFrame({
    "Invalid_Employee_ID": invalid_employee_id,
    "Duplicate_Employee_ID": duplicate_employee_id,
    "Invalid_Employee_Name": invalid_employee_name,
    "Invalid_Age": invalid_age,
    "Invalid_Salary": invalid_salary,
    "Invalid_Department": invalid_department,
    "Invalid_Joining_Date": invalid_joining_date
})

print("\nValidation Report:")
print(validation_report)

# Step 10: Display rows that need investigation
any_validation_issue = validation_report.any(axis=1)

print("\nRecords Requiring Investigation:")
print(df.loc[any_validation_issue])

# Step 11: Count validation problems by rule
print("\nNumber of Problems for Each Rule:")
print(validation_report.sum())
```
