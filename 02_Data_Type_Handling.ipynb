# Data Type Handling

## 1. What is Data Type Handling?

Data type handling is the process of identifying, correcting, and converting the data types of columns so that they represent information correctly.

In a dataset, values may be stored using incorrect data types. For example, a column containing numerical values may be stored as strings. This can cause errors during calculations, analysis, and Machine Learning preprocessing.

Data type handling helps ensure that each column uses an appropriate data type.

## 2. Why is Data Type Handling Important?

* It helps perform mathematical calculations correctly.
* It prevents errors during data analysis.
* It helps identify invalid values.
* It ensures that dates are stored in the correct format.
* It helps distinguish numerical and categorical features.
* It prepares data for Machine Learning preprocessing.
* It improves data consistency and reliability.

## 3. Types of Data

### Numerical Data

Numerical data represents measurable quantities or counts.

Examples:

* Age
* Salary
* MonthlyCharges
* Tenure

Numerical data can be divided into:

* **Discrete:** Countable values, such as the number of children.
* **Continuous:** Measurable values, such as height or weight.

### Categorical Data

Categorical data represents groups or labels.

Examples:

* Gender
* Contract type
* Payment method
* Customer status

Categorical data may be nominal or ordinal.

* **Nominal:** Categories without a natural order, such as payment methods.
* **Ordinal:** Categories with a meaningful order, such as Low, Medium, and High.

### Boolean Data

Boolean data represents two logical values: `True` and `False`.

Example:

* IsActive
* HasSubscription
* IsChurned

### Date/Time Data

Date/time data represents dates, times, or timestamps.

Examples:

* DateOfBirth
* JoiningDate
* TransactionDate

The `datetime` data type allows date-based calculations and operations.

### String Data

String data contains text.

Examples:

* CustomerName
* Address
* CustomerID

An important point is that some strings contain digits but are not numerical measurements.

For example, a customer ID such as `"1001"` should usually remain an identifier rather than being treated as a numerical feature.

## 4. Identifying Data Types

Pandas provides methods to inspect column data types.

* `df.dtypes`: Displays the data type of each column.
* `df.info()`: Displays column names, data types, and non-null counts.
* `df.select_dtypes()`: Selects columns based on their data types.

## 5. Converting Data Types

### astype()

The `astype()` method converts a column into a specified data type.

Example:

`df["Age"] = df["Age"].astype(int)`

Use it when the values can be safely represented by the target type.

### to_numeric()

The `pd.to_numeric()` function converts values into numerical data.

Example:

`pd.to_numeric(df["Salary"], errors="coerce")`

With `errors="coerce"`, invalid values become `NaN`.

### to_datetime()

The `pd.to_datetime()` function converts values into date/time data.

Example:

`pd.to_datetime(df["JoiningDate"], errors="coerce")`

With `errors="coerce"`, unrecognized date values become `NaT`.

## 6. Handling Invalid Conversions

Some values cannot be converted into the requested data type.

Examples:

* `"unknown"` cannot be directly converted into a number.
* `"not-a-date"` cannot be interpreted as a valid date.
* A decimal value cannot always be converted to an integer without losing information.

Invalid values should be identified and investigated before deciding whether to correct, retain, or replace them.

## 7. Common Data Type Conversions

* String → Numeric
* String → Date/Time
* Numeric → Categorical
* Integer → Float
* String → Boolean

The correct conversion depends on what the data represents and how it will be used.

## Conclusion

Data type handling is an important preprocessing step because incorrect data types can cause calculation errors, hide data-quality problems, and affect later transformations.

Always inspect the data before converting it, and validate the result after conversion.


# Demonstration Data Type Handling

## Simple Example

Consider the following customer dataset.

| CustomerID | Age | Salary  | JoiningDate  | IsActive |
| ---------- | --- | ------- | ------------ | -------- |
| C101       | 25  | 30000   | 2024-01-15   | True     |
| C102       | 30  | 45000   | 2023-06-10   | False    |
| C103       | 28  | unknown | invalid-date | True     |

### Problems Identified

1. Age may be stored as a string instead of a numerical value.
2. Salary contains the text `"unknown"` instead of a valid number.
3. JoiningDate contains an invalid date.
4. CustomerID contains digits but represents an identifier, not a measurable quantity.
5. IsActive represents a Boolean condition.

### How Can We Address These Problems?

* Convert Age into a suitable numerical type.
* Convert Salary using `pd.to_numeric()`.
* Convert JoiningDate using `pd.to_datetime()`.
* Keep CustomerID as a string because it identifies customers.
* Convert IsActive into a Boolean type if it is not already stored that way.
* Investigate invalid values instead of silently assuming they are correct.

## Real-World Business Example

A company receives customer information from different systems.

One system stores salaries as numbers, while another stores them as strings. Some date fields may also contain inconsistent formats.

If the data types are incorrect, the company may encounter problems when calculating average salaries, measuring customer tenure, or analyzing joining dates.

Data type handling makes these operations more reliable.

## AI/ML Use Case

In a customer churn prediction project:

* Tenure and MonthlyCharges should generally be numerical.
* Contract and PaymentMethod are categorical.
* CustomerID is usually an identifier and should not automatically be used as a predictive feature.
* JoiningDate should be converted into a date/time type if date-based features are needed.

Correct data types help ensure that the appropriate preprocessing technique is applied to each feature.

For example, numerical features may require imputation and scaling, while categorical features may require imputation and encoding.


```python
import pandas as pd

# Create a sample dataset containing different data types
data = {
    "CustomerID": ["C101", "C102", "C103", "C104"],
    "Age": ["25", "30", "28", "35"],
    "Salary": ["30000", "45000", "unknown", "50000"],
    "JoiningDate": [
        "2024-01-15",
        "2023-06-10",
        "invalid-date",
        "2022-12-01"
    ],
    "IsActive": ["True", "False", "True", "False"],
    "SatisfactionScore": [1, 2, 3, 4]
}

df = pd.DataFrame(data)

# Step 1: Display the original dataset
print("ORIGINAL DATASET")
print(df)

# Step 2: Inspect the original data types
print("\nORIGINAL DATA TYPES")
print(df.dtypes)

# Step 3: Convert Age from string to integer
df["Age"] = df["Age"].astype(int)

# Step 4: Convert Salary from string to numeric
df["Salary"] = pd.to_numeric(
    df["Salary"],
    errors="coerce"
)

# Step 5: Convert JoiningDate from string to datetime
df["JoiningDate"] = pd.to_datetime(
    df["JoiningDate"],
    errors="coerce"
)

# Step 6: Convert IsActive from string to Boolean
df["IsActive"] = df["IsActive"].map({
    "True": True,
    "False": False
}).astype("boolean")

# Step 7: Convert SatisfactionScore from integer to float
df["SatisfactionScore"] = df["SatisfactionScore"].astype(float)

# Step 8: Display the converted dataset
print("\nDATASET AFTER TYPE CONVERSION")
print(df)

# Step 9: Inspect the updated data types
print("\nUPDATED DATA TYPES")
print(df.dtypes)

# Step 10: Identify invalid or missing values created by conversion
print("\nMISSING VALUES AFTER CONVERSION")
print(df.isnull().sum())

# Step 11: Display rows with invalid Salary or JoiningDate
print("\nROWS WITH INVALID SALARY")
print(df[df["Salary"].isna()])

print("\nROWS WITH INVALID JOINING DATE")
print(df[df["JoiningDate"].isna()])
```
