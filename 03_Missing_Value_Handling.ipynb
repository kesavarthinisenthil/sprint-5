# Notebook 3 – Missing Value Handling

## 1. What Are Missing Values?

Missing values occur when information is not available for one or more entries in a dataset.

Missing values may appear as `NaN`, `None`, or `NaT` in Pandas. Sometimes, missing information is represented using empty strings or placeholders such as `"Unknown"` or `"N/A"`.

Example:

| CustomerID | Age | MonthlyCharges |
| ---------- | --: | -------------: |
| 101        |  25 |            500 |
| 102        | NaN |            700 |
| 103        |  35 |            NaN |
| 104        |  40 |            800 |

Here, Customer 102 has a missing age, and Customer 103 has missing monthly charges.

## 2. Why Do Missing Values Occur?

Missing values may occur because of:

* Customers not providing information.
* Errors during data collection.
* Failed database connections.
* Incomplete surveys.
* Sensor or system failures.
* Incorrect data entry.
* Information not being applicable to a particular record.

Before handling missing values, it is important to understand why the information is missing.

## 3. Types of Missing Data

### MCAR – Missing Completely At Random

MCAR occurs when the probability that a value is missing is unrelated to observed or unobserved data.

Example: Some customer records are missing because a data collection system randomly failed.

When to consider it:

* When missingness appears unrelated to the available information.

Limitation:

* It can be difficult to establish that data is truly MCAR.

### MAR – Missing At Random

MAR occurs when missingness depends on other observed variables, but not on the missing value itself after accounting for those observed variables.

Example: Customers in a particular age group are less likely to complete a salary field, and age is available in the dataset.

When to consider it:

* When missingness can reasonably be explained by observed variables.

Limitation:

* The assumption must be evaluated using domain knowledge and appropriate analysis.

### MNAR – Missing Not At Random

MNAR occurs when missingness depends on the missing value itself or on other unobserved information.

Example: Customers with particularly high incomes may be less willing to disclose their income.

When to consider it:

* When the missingness mechanism may depend on the missing value or unobserved factors.

Limitation:

* It is difficult to identify and handle without additional information or assumptions.

## 4. Detecting Missing Values

Pandas provides several useful methods.

### isnull() and isna()

These methods identify missing values.

Example:

`df.isnull()`

### sum()

Counts missing values in each column.

Example:

`df.isnull().sum()`

### Missing Value Percentage

The percentage of missing values in each column can be calculated using:

`df.isnull().mean() * 100`

This helps identify columns with a large proportion of missing information.

## 5. Dropping Rows

Dropping rows removes records containing missing values.

Example:

`df.dropna()`

When to use:

* Only a small number of rows are missing important information.
* Removing them is appropriate for the analysis.
* The resulting dataset remains sufficiently representative.

Advantages:

* Simple to implement.
* Does not introduce artificial imputed values.

Limitations:

* Reduces the dataset size.
* May introduce bias if missingness is systematic.
* Can discard useful information from other columns.

## 6. Dropping Columns

Dropping columns removes entire features containing missing values.

Example:

`df.drop(columns=["ColumnName"])`

When to use:

* A feature has a large proportion of missing values.
* The feature is not essential to the task.
* Its information cannot be recovered or used reliably.

Advantages:

* Simplifies the dataset.
* Avoids unreliable imputation of an unsuitable feature.

Limitations:

* May discard useful predictive information.
* A high missing-value percentage alone does not justify removing a column.

## 7. Mean Imputation

Mean imputation replaces missing numerical values with the arithmetic mean of the available values.

Example:

`df["Age"] = df["Age"].fillna(df["Age"].mean())`

When to use:

* The feature is numerical.
* The distribution is reasonably symmetric.
* Extreme values are not strongly affecting the mean.

Advantages:

* Simple and fast.
* Preserves the number of rows.

Limitations:

* Sensitive to outliers.
* Reduces the observed variance.
* Can distort relationships between features.

## 8. Median Imputation

Median imputation replaces missing numerical values with the median of the available observations.

Example:

`df["Age"] = df["Age"].fillna(df["Age"].median())`

When to use:

* Numerical data is skewed.
* Outliers are present.
* A robust measure of central tendency is appropriate.

Advantages:

* Less affected by extreme values than the mean.
* Simple to implement.

Limitations:

* Reduces variability.
* May distort relationships between variables.
* Does not account for other features that could explain the missing value.

## 9. Mode Imputation

Mode imputation replaces missing values with the most frequently occurring value.

Example:

`df["Contract"] = df["Contract"].fillna(df["Contract"].mode()[0])`

When to use:

* Categorical features contain missing values.
* The most common category is a reasonable replacement.

Advantages:

* Easy to implement.
* Preserves existing category labels.

Limitations:

* Can make the most frequent category appear more common.
* May introduce bias.
* May be unsuitable when missingness itself carries useful information.

## 10. Constant Value Imputation

Constant imputation replaces missing values with a predefined value.

Examples:

* `"Unknown"` for an appropriate categorical feature.
* `0` when zero has a valid and meaningful interpretation.
* A predefined sentinel value when the model and business rules support it.

When to use:

* A missing category should remain identifiable.
* A meaningful business-defined replacement exists.

Advantages:

* Simple and predictable.
* Can preserve information that a value was missing.

Limitations:

* A constant such as zero may have a misleading meaning.
* A new category may require handling by the encoding method.
* The chosen constant can distort numerical distributions.

## 11. Forward Fill

Forward fill replaces a missing value with the most recent preceding non-missing value.

Example:

`df["Value"] = df["Value"].ffill()`

When to use:

* Ordered or time-series data.
* The previous observation is a reasonable estimate of the missing value.

Advantages:

* Preserves the sequence of observations.
* Useful for some time-series and event data.

Limitations:

* Can propagate outdated values.
* Results depend on row order.
* Should not be applied blindly to unrelated customer records.

## 12. Backward Fill

Backward fill replaces a missing value with the next available non-missing value.

Example:

`df["Value"] = df["Value"].bfill()`

When to use:

* Ordered data where the next observation is appropriate.
* Situations where future observations are legitimately available.

Advantages:

* Simple for suitable ordered datasets.
* Can fill gaps using the next known observation.

Limitations:

* Can introduce future information into a prediction workflow.
* May cause data leakage when used improperly.
* Can propagate an inappropriate value backward.

## 13. Interpolation

Interpolation estimates missing numerical values using surrounding observations.

Example:

`df["Value"] = df["Value"].interpolate()`

When to use:

* Ordered numerical data.
* Time series or measurements that change smoothly.

Advantages:

* Uses information from neighboring observations.
* Can produce more realistic estimates than a constant replacement for smooth sequences.

Limitations:

* Depends on assumptions about how values change.
* May perform poorly when changes are abrupt.
* Some interpolation methods use future observations and may be unsuitable for real-time prediction.

## 14. Group-Based Imputation

Group-based imputation replaces missing values using statistics calculated within relevant groups.

Example: Fill missing customer tenure using the median tenure for customers with the same contract type.

When to use:

* Groups have meaningful differences.
* Group membership provides useful information about the missing feature.

Advantages:

* Accounts for differences between groups.
* Can be more appropriate than a single global statistic.

Limitations:

* Small groups may produce unreliable estimates.
* Some groups may have no observed values.
* Group definitions must be justified by the data and business context.

## 15. KNN Imputation

K-Nearest Neighbors imputation estimates missing numerical values using similar observations.

It identifies nearby records based on available features and uses information from those neighbors to estimate missing values.

When to use:

* Similar records are expected to have similar values.
* The dataset has suitable features for measuring similarity.

Advantages:

* Can account for relationships between features.
* Can produce more individualized estimates than a global mean or median.

Limitations:

* Can be computationally expensive.
* Sensitive to feature scaling and the choice of neighbors.
* May perform poorly with noisy data or unsuitable distance measures.

Scikit-learn provides `KNNImputer` for numerical missing-value imputation.

## 16. Iterative Imputation

Iterative imputation estimates missing values by modeling each incomplete feature using other features.

The process is repeated across features until the estimates stabilize or a stopping condition is reached.

When to use:

* Features have meaningful relationships.
* A model-based imputation method is appropriate.

Advantages:

* Can use relationships among multiple features.
* May capture more complex patterns than simple imputation.

Limitations:

* More computationally expensive.
* Results depend on the models and assumptions used.
* May introduce unreliable estimates if the relationships are weak.

Scikit-learn provides `IterativeImputer` for this approach.

## 17. Important Rules for Missing Value Handling

1. Inspect missing values before selecting a technique.
2. Investigate why values are missing whenever possible.
3. Do not automatically replace every numerical missing value with the mean.
4. Consider the data type, distribution, business rules, and purpose of the feature.
5. Compare the advantages and limitations of alternative techniques.
6. Record how missing values were handled.
7. For predictive modeling, split the data before fitting learned imputation strategies.
8. Fit imputers on training data only, then use the fitted imputers to transform validation and test data.
9. Consider whether missingness itself carries useful information.
10. Validate the results after imputation.

## Conclusion

Missing value handling is an important part of data preprocessing. The appropriate technique depends on why values are missing, the type and distribution of the feature, the business context, and the Machine Learning objective.

The goal is not simply to eliminate missing values. The goal is to prepare reliable data while minimizing information loss, bias, and data leakage.


# Demonstration Missing Value Handling

## Simple Example

Consider the following customer dataset.

| CustomerID | Age | MonthlyCharges | Contract |
| ---------- | --: | -------------: | -------- |
| 101        |  25 |            500 | Monthly  |
| 102        | NaN |            700 | Yearly   |
| 103        |  35 |            NaN | Monthly  |
| 104        |  40 |            800 | Yearly   |
| 105        |  30 |            600 | NaN      |

### Problems Identified

1. Customer 102 has a missing age.
2. Customer 103 has missing monthly charges.
3. Customer 105 has a missing contract type.

### Analysis

* Age is numerical.
* MonthlyCharges is numerical and contains an extreme value risk that should be checked before choosing an imputation method.
* Contract is categorical.
* The reason for each missing value is not yet known.

### Technique Selected

For this demonstration:

* **Age:** Median imputation.
* **MonthlyCharges:** Median imputation.
* **Contract:** Constant imputation using `"Unknown"`.

### Why Were These Techniques Selected?

**Median imputation:** The median is less sensitive to extreme values than the mean and is a useful baseline for numerical features.

**Constant imputation:** `"Unknown"` keeps the missing contract information identifiable rather than assuming that the customer has the most common contract type.

These choices are examples, not universal rules. A real dataset requires further investigation before finalizing the approach.

## Real-World Business Example

A telecommunications company collects customer information to understand customer behavior and predict churn.

Some customers have missing tenure, monthly charges, or contract information.

If these missing values are not handled appropriately, the company may lose useful customer records or introduce misleading information into its analysis.

The company should investigate the cause of the missing values and select an appropriate technique for each feature.

## AI/ML Use Case

For customer churn prediction:

* Numerical missing values may be imputed using the median or another suitable method.
* Categorical missing values may be represented by an `"Unknown"` category or imputed using another justified strategy.
* KNN or iterative imputation may be useful when relationships among features provide reliable information.
* Time-series methods may be appropriate for sequential measurements.

For predictive modeling, the imputation strategy must be learned from the training data only. The same fitted strategy is then applied to validation and test data.

**Important:** Filling missing values does not guarantee better model performance. The chosen method should be justified and evaluated as part of the complete Machine Learning workflow.


```python
import pandas as pd
import numpy as np

from sklearn.impute import KNNImputer
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer

# Create a sample customer dataset
data = {
    "CustomerID": [101, 102, 103, 104, 105, 106],
    "Age": [25, np.nan, 35, 40, 30, np.nan],
    "MonthlyCharges": [500, 700, np.nan, 800, 600, 750],
    "Contract": ["Monthly", "Yearly", "Monthly", np.nan, "Yearly", "Monthly"],
    "Tenure": [2, 5, np.nan, 8, 4, 7],
    "Region": ["South", "North", "South", "North", np.nan, "South"]
}

df = pd.DataFrame(data)

# Step 1: Display the original dataset
print("ORIGINAL DATASET")
print(df)

# Step 2: Identify missing values
print("\nMISSING VALUE COUNTS")
print(df.isnull().sum())

# Step 3: Calculate missing-value percentages
print("\nMISSING VALUE PERCENTAGE")
print((df.isnull().mean() * 100).round(2))

# Step 4: Demonstrate dropping rows
rows_dropped = df.dropna()

print("\nDATASET AFTER DROPPING ROWS WITH MISSING VALUES")
print(rows_dropped)

# Step 5: Demonstrate dropping a column
# Use a copy so the original dataset remains unchanged
column_dropped = df.drop(columns=["Region"])

print("\nDATASET AFTER DROPPING REGION COLUMN")
print(column_dropped)

# Step 6: Mean imputation
mean_df = df.copy()
mean_df["Age"] = mean_df["Age"].fillna(mean_df["Age"].mean())

print("\nMEAN IMPUTATION")
print(mean_df[["CustomerID", "Age"]])

# Step 7: Median imputation
median_df = df.copy()
median_df["MonthlyCharges"] = median_df["MonthlyCharges"].fillna(
    median_df["MonthlyCharges"].median()
)

print("\nMEDIAN IMPUTATION")
print(median_df[["CustomerID", "MonthlyCharges"]])

# Step 8: Mode imputation
mode_df = df.copy()
mode_df["Contract"] = mode_df["Contract"].fillna(
    mode_df["Contract"].mode()[0]
)

print("\nMODE IMPUTATION")
print(mode_df[["CustomerID", "Contract"]])

# Step 9: Constant-value imputation
constant_df = df.copy()
constant_df["Contract"] = constant_df["Contract"].fillna("Unknown")

print("\nCONSTANT IMPUTATION")
print(constant_df[["CustomerID", "Contract"]])

# Step 10: Forward fill and backward fill
# Demonstration only: row order must have a valid meaning
sequence = pd.Series([10, np.nan, 30, np.nan, 50])

print("\nORIGINAL SEQUENCE")
print(sequence)

print("\nFORWARD FILL")
print(sequence.ffill())

print("\nBACKWARD FILL")
print(sequence.bfill())

# Step 11: Interpolation
print("\nINTERPOLATION")
print(sequence.interpolate())

# Step 12: Group-based imputation
group_df = df.copy()

group_df["Age"] = group_df.groupby("Contract")["Age"].transform(
    lambda group: group.fillna(group.median())
)

print("\nGROUP-BASED IMPUTATION")
print(group_df[["CustomerID", "Contract", "Age"]])

# Step 13: KNN imputation
numerical_columns = ["Age", "MonthlyCharges", "Tenure"]

knn_df = df.copy()

knn_imputer = KNNImputer(n_neighbors=2)

knn_df[numerical_columns] = knn_imputer.fit_transform(
    knn_df[numerical_columns]
)

print("\nKNN IMPUTATION")
print(knn_df[["CustomerID"] + numerical_columns])

# Step 14: Iterative imputation
iterative_df = df.copy()

iterative_imputer = IterativeImputer(
    max_iter=10,
    random_state=42
)

iterative_df[numerical_columns] = iterative_imputer.fit_transform(
    iterative_df[numerical_columns]
)

print("\nITERATIVE IMPUTATION")
print(iterative_df[["CustomerID"] + numerical_columns])

# Step 15: Compare missing values before and after
print("\nMISSING VALUES BEFORE IMPUTATION")
print(df.isnull().sum())

print("\nMISSING VALUES AFTER SIMPLE IMPUTATION")
print(constant_df.isnull().sum())
```
