# Data Preprocessing Fundamentals

## 1. What is Data Preprocessing?

Data preprocessing is the process of converting raw data into a clean, consistent, and suitable format for Machine Learning.

Raw data may contain missing values, duplicate records, incorrect data types, outliers, and inconsistent values. These problems can affect the quality of analysis and the performance of Machine Learning models.

Data preprocessing helps identify and correct these problems before training a model.

### Why is Data Preprocessing Required?

* To handle missing values.
* To remove unnecessary duplicate records.
* To correct incorrect data types.
* To identify and handle outliers.
* To convert categorical values into numerical representations when required.
* To scale numerical features when necessary.
* To improve data quality and consistency.
* To prepare data for Machine Learning algorithms.

## 2. Raw Data vs Clean Data

**Raw Data:** Data collected from a source before the necessary cleaning and transformations have been applied.

**Clean Data:** Data that has been checked and processed to address relevant quality issues and is suitable for the intended analysis or Machine Learning task.

Example:

| Customer | Age | MonthlyCharges |
| -------- | --: | -------------: |
| A        |  25 |            500 |
| B        | NaN |            700 |
| C        |  35 |            600 |
| C        |  35 |            600 |

The raw dataset contains a missing age and a duplicate record.

After appropriate preprocessing, the missing age may be imputed using a suitable method, and the duplicate may be removed if it represents an accidental duplicate.

## 3. Data Quality

Data quality describes how suitable data is for its intended purpose.

Important dimensions of data quality include:

### Data Consistency

Data is consistent when the same information follows the same format and rules across records.

Example: A category should not be represented as both `Male` and `male` when they are intended to mean the same thing.

### Data Completeness

Data is complete when the required information is available.

Example: A customer record with a missing age has incomplete information if age is required for the analysis.

### Data Accuracy

Data is accurate when it correctly represents the real-world information it is intended to describe.

Example: A customer's actual age is 25, but the dataset records it as 250.

### Data Validity

Data is valid when it follows the expected format, type, and business rules.

Example: A percentage should be between 0 and 100 when that is the defined business rule.

### Data Integrity

Data integrity means that data remains reliable and maintains its correctness and consistency throughout collection, storage, and processing.

Example: Every order should reference a valid customer ID in the customer table.

## 4. Data Leakage

Data leakage occurs when information that would not legitimately be available at prediction time influences model training.

This can produce misleading evaluation results and make a model appear more effective than it really is.

Example: A model predicts whether a customer will leave a company, but one feature contains information recorded only after the customer has left.

That feature leaks future information into the prediction process.

## 5. Training, Validation, and Test Data

### Training Data

Training data is used to teach the Machine Learning model patterns in the dataset.

### Validation Data

Validation data is used to compare models, tune hyperparameters, and make model-selection decisions.

### Test Data

Test data is used for the final evaluation of the selected model on data that was not used to fit the model or select its settings.

A common split is 70% training data, 15% validation data, and 15% test data. The appropriate split depends on the dataset and problem.

## 6. Preprocessing Pipeline

A preprocessing pipeline is an organized sequence of data-processing steps that can be applied consistently.

Example workflow:

1. Load the dataset.
2. Inspect the dataset.
3. Identify data-quality problems.
4. Handle missing values.
5. Handle duplicate records when appropriate.
6. Validate data.
7. Handle outliers when necessary.
8. Encode categorical features.
9. Scale numerical features when necessary.
10. Prepare the data for Machine Learning.

For predictive modeling, split the data appropriately before fitting preprocessing transformations. Fit learned transformations using training data only, then apply them to validation and test data.

## Conclusion

Data preprocessing is an essential stage of the Machine Learning workflow. It helps address data-quality problems, prevent data leakage, and prepare reliable input for model training and evaluation.


# Demonstration of Data Preprocessing

## Simple Example

Consider a dataset containing customer information.

| CustomerID | Age | MonthlyCharges | Contract |
| ---------- | --: | -------------: | -------- |
| 101        |  25 |            500 | Monthly  |
| 102        | NaN |            700 | Yearly   |
| 103        |  35 |            600 | Monthly  |
| 103        |  35 |            600 | Monthly  |
| 104        | 250 |            800 | Yearly   |

### Problems Identified

1. Customer 102 has a missing age.
2. Customer 103 appears twice with identical information.
3. Customer 104 has an age of 250, which is suspicious and requires validation.
4. Contract is categorical data.

### How Can We Address These Problems?

* **Missing value:** Investigate why age is missing and select an appropriate imputation method if needed.
* **Duplicate record:** Verify whether the repeated record is accidental before removing it.
* **Suspicious age:** Check the source and business rules before deciding whether to correct, retain, or exclude it.
* **Categorical feature:** Encode Contract if the selected Machine Learning algorithm requires numerical input.

## Real-World Business Example

A telecommunications company wants to predict which customers are likely to leave its service.

Its customer dataset may contain missing tenure values, inconsistent contract names, duplicate customer records, and incorrect numerical values.

If these problems are not addressed appropriately, the model may learn misleading patterns and produce unreliable predictions.

Data preprocessing helps the company prepare trustworthy data for churn prediction.

## AI/ML Use Case

In a customer churn prediction model:

* Missing values can be handled using suitable imputation techniques.
* Duplicate records can be investigated and resolved.
* Categorical features can be encoded.
* Numerical features can be scaled when required by the algorithm.
* Data can be split into training, validation, and test sets.
* A preprocessing pipeline can help apply transformations consistently.

**Important:** Not every problem requires deleting data. Each decision should be based on data quality, business rules, and the Machine Learning objective.


```python
import pandas as pd
import numpy as np

# Create a sample raw dataset
data = {
    "CustomerID": [101, 102, 103, 103, 104],
    "Age": [25, np.nan, 35, 35, 250],
    "MonthlyCharges": [500, 700, 600, 600, 800],
    "Contract": ["Monthly", "Yearly", "Monthly", "Monthly", "Yearly"]
}

df = pd.DataFrame(data)

# Display the raw dataset
print("RAW DATASET")
print(df)

# 1. Inspect dataset dimensions and data types
print("\nDataset Shape:", df.shape)

print("\nData Types:")
print(df.dtypes)

# 2. Identify missing values
print("\nMissing Values:")
print(df.isnull().sum())

# 3. Identify duplicate records
print("\nDuplicate Records:", df.duplicated().sum())

# 4. Check suspicious ages using a business rule
print("\nRecords with suspicious ages:")
print(df[(df["Age"] < 0) | (df["Age"] > 120)])

# 5. Remove exact duplicate records
clean_df = df.drop_duplicates().copy()

# 6. Impute missing ages using the median
median_age = clean_df["Age"].median()
clean_df["Age"] = clean_df["Age"].fillna(median_age)

# 7. Validate the age range
# Flag suspicious values rather than silently changing them
clean_df["Age_Valid"] = clean_df["Age"].between(0, 120)

# Display the processed dataset
print("\nPROCESSED DATASET")
print(clean_df)

# 8. Compare before and after
print("\nBEFORE PREPROCESSING")
print("Shape:", df.shape)
print("Missing values:", df.isnull().sum().sum())
print("Duplicate records:", df.duplicated().sum())

print("\nAFTER PREPROCESSING")
print("Shape:", clean_df.shape)
print("Missing values:", clean_df.isnull().sum().sum())
print("Duplicate records:", clean_df.duplicated().sum())
print("Invalid ages:", (~clean_df["Age_Valid"]).sum())
```
