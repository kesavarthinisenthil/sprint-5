# Encoding Categorical Variables

## 1. What Is Categorical Encoding?

Categorical encoding is the process of converting categorical data into a numerical representation that Machine Learning algorithms can use.

Categorical data contains labels or groups instead of ordinary numerical measurements.

Examples:

* Gender: Female, Male
* Department: IT, HR, Finance
* Education: School, Bachelor's, Master's, PhD
* Payment Method: Cash, Card, UPI
* Customer Status: Active, Inactive

Many Machine Learning algorithms require numerical input. Encoding helps represent categorical values numerically while preserving their meaning as much as possible.

**Example:**

Original data:

| Customer | Payment Method |
| -------- | -------------- |
| A        | Cash           |
| B        | UPI            |
| C        | Card           |

After one-hot encoding:

| Customer | Cash | UPI | Card |
| -------- | ---: | --: | ---: |
| A        |    1 |   0 |    0 |
| B        |    0 |   1 |    0 |
| C        |    0 |   0 |    1 |

The categorical labels are converted into numerical columns.

**Important:** Encoding is not simply about converting text into numbers. We must choose a technique that represents the meaning of the category correctly.

## 2. Types of Categorical Variables

Before selecting an encoding method, identify the type of categorical variable.

### A. Nominal Variables

Nominal variables represent categories that have no natural order.

Examples:

* Department: IT, HR, Finance
* Color: Red, Blue, Green
* City: Chennai, Coimbatore, Madurai
* Payment Method: Cash, Card, UPI

There is no meaningful ranking between these categories.

**Recommended techniques:**

* One-Hot Encoding
* Dummy Encoding
* Frequency Encoding
* Target Encoding, when justified and implemented without leakage

**When to use:** One-hot encoding is a good starting point for nominal variables with a manageable number of categories.

### B. Ordinal Variables

Ordinal variables represent categories that have a meaningful order.

Examples:

* Education: School < Bachelor's < Master's < PhD
* Satisfaction: Low < Medium < High
* Clothing Size: Small < Medium < Large

The categories have an order, but the distances between them may not be equal.

**Recommended technique:** Ordinal Encoding.

For example:

| Education  | Encoded Value |
| ---------- | ------------: |
| School     |             0 |
| Bachelor's |             1 |
| Master's   |             2 |
| PhD        |             3 |

**When to use:** Use ordinal encoding when the order is known and meaningful.

**Risk:** The encoded numbers may imply equal spacing between categories. This may not accurately represent the real-world differences between them.

### C. Binary Variables

Binary variables have exactly two possible categories.

Examples:

* Yes / No
* True / False
* Active / Inactive
* Approved / Rejected

A binary variable can often be represented using 0 and 1.

Example:

| Approved | Encoded Value |
| -------- | ------------: |
| No       |             0 |
| Yes      |             1 |

**When to use:** Use binary encoding for two-category variables when the mapping is clearly defined.

**Important:** The choice of which category becomes 0 or 1 should be documented. If the variable is a target label, the mapping must be consistent with the model's intended interpretation.

## 3. Label Encoding

Label encoding assigns an integer to each distinct category.

Example:

| Department | Encoded Value |
| ---------- | ------------: |
| Finance    |             0 |
| HR         |             1 |
| IT         |             2 |

**When to use:**

* Encoding class labels for a classification target.
* Encoding categories for algorithms that explicitly support categorical features or when integer codes have a suitable interpretation.
* Cases where an integer representation is needed without assuming the values represent a meaningful ranking.

**Risks:**

* Some algorithms may treat the numbers as ordered or as numerical distances.
* A model may incorrectly interpret IT as greater than HR because 2 is greater than 1.
* The numerical ordering may be arbitrary.

**Recommendation:** Avoid arbitrary integer encoding for nominal input features when the chosen algorithm treats the encoded numbers as continuous measurements.

## 4. Ordinal Encoding

Ordinal encoding assigns numerical values according to the natural order of categories.

Example:

| Satisfaction | Encoded Value |
| ------------ | ------------: |
| Low          |             1 |
| Medium       |             2 |
| High         |             3 |

**When to use:**

* Customer satisfaction levels.
* Education levels.
* Product ratings grouped into ordered categories.
* Other categories with a meaningful, known order.

**Risks:**

* The order must be defined correctly.
* The numerical gaps do not necessarily represent equal real-world differences.
* Incorrect ordering can mislead a model.

Use an explicit category order rather than allowing software to guess the order.

## 5. One-Hot Encoding

One-hot encoding creates a separate binary column for each category.

Example:

Original data:

| Color |
| ----- |
| Red   |
| Blue  |
| Green |
| Red   |

Encoded data:

| Color_Blue | Color_Green | Color_Red |
| ---------: | ----------: | --------: |
|          0 |           0 |         1 |
|          1 |           0 |         0 |
|          0 |           1 |         0 |
|          0 |           0 |         1 |

Each row receives a 1 in the column representing its category and 0 in the other category columns.

**When to use:**

* Nominal variables with a manageable number of categories.
* Categories without a natural ranking.
* Linear models and many other algorithms that require numerical input.

**Advantages:**

* Does not impose an arbitrary ranking.
* Makes category membership explicit.
* Works well for many low-cardinality categorical features.

**Risks:**

* Can create many columns when there are many categories.
* May increase memory usage and model complexity.
* If all categories are included in a linear model with an intercept, perfect multicollinearity can occur. Dropping one category can help address this in suitable models.

## 6. Dummy Encoding

Dummy encoding is similar to one-hot encoding but generally drops one category to use it as a reference category.

Example:

Original categories: Red, Blue, Green.

Dummy-encoded data:

| Color_Blue | Color_Green |
| ---------: | ----------: |
|          0 |           0 |
|          1 |           0 |
|          0 |           1 |
|          0 |           0 |

Here, Red is the reference category.

When both dummy columns are 0, the row represents Red.

**When to use:**

* Linear regression and other models where a reference category is useful.
* Statistical analysis that compares categories against a baseline.

**Risk:** The interpretation depends on the reference category. Dropping a category is not always necessary for every model, so the choice depends on the algorithm and purpose.

## 7. Frequency Encoding

Frequency encoding replaces each category with its frequency or proportion in the dataset.

Example:

| Department | Frequency |
| ---------- | --------: |
| IT         |         4 |
| HR         |         2 |
| Finance    |         1 |
| IT         |         4 |

If IT occurs four times, its frequency value is 4.

Alternatively, we could use proportions:

* IT: 4/7
* HR: 2/7
* Finance: 1/7

**When to use:**

* High-cardinality features with many unique categories.
* Situations where category frequency provides useful information.
* Cases where one-hot encoding would create too many columns.

**Advantages:**

* Produces a single numerical column.
* Helps control feature dimensionality.

**Risks:**

* Different categories can have the same frequency.
* Frequency does not necessarily indicate that categories have similar meaning.
* Frequency values may change between training and production data.

Calculate frequency mappings using training data and define a policy for unseen categories.

## 8. Target Encoding

Target encoding replaces each category with a statistic calculated from the target variable.

For a binary classification problem, a category might be represented by the average target value for that category.

Example:

Suppose the target is whether a customer purchases a product:

* IT customers: purchase rate = 0.60
* HR customers: purchase rate = 0.40
* Finance customers: purchase rate = 0.20

The department could be represented using these purchase rates.

**When to use:**

* High-cardinality categorical features.
* Problems where category-specific target information may be useful.
* Cases where one-hot encoding would create an excessive number of features.

**Risks:**

* Target leakage if target statistics are calculated using validation or test labels.
* Overfitting, particularly for categories with very few observations.
* Unstable estimates for rare categories.

**Important precautions:**

* Fit the encoder using training data only.
* Use cross-fitting or out-of-fold target encoding for training rows.
* Consider smoothing to reduce overfitting for rare categories.
* Do not calculate category target statistics from the validation or test targets.
* Use a defined fallback for unseen categories.

Target encoding must be implemented carefully because it uses information from the target variable.

## 9. Handling Unknown Categories

An unknown category is a value that appears during validation, testing, or production but was not present when the encoder was fitted.

Example:

Training categories:

* IT
* HR
* Finance

New data contains:

* Marketing

Marketing is an unknown category for the fitted encoder.

Possible strategies include:

1. **Ignore unknown categories:** With one-hot encoding, the unseen category may be represented by zeros across the learned category columns.
2. **Use an unknown category:** Map unseen values to a predefined category such as `Unknown`, if the encoder was configured to recognize it during training.
3. **Group rare categories:** Combine infrequent categories into an `Other` category when justified.
4. **Reject or flag unexpected values:** Appropriate when categories must follow strict business rules.

**Recommendation:** Choose a consistent policy for unseen categories before deploying a model.

## 10. High-Cardinality Categories

High-cardinality features contain many unique categories.

Examples:

* Postal codes.
* Product IDs.
* Customer IDs.
* Thousands of product names.
* Large numbers of job titles.

For example, a product column with 5,000 unique products could create approximately 5,000 columns with ordinary one-hot encoding.

Possible approaches include:

* Grouping rare categories into `Other`.
* Frequency encoding.
* Carefully implemented target encoding.
* Feature hashing, when suitable.
* Domain-specific feature engineering.

**Important:** Customer IDs and product IDs are identifiers, not necessarily meaningful predictive features. Encoding them does not automatically make them useful. Assess whether they capture genuine information or cause overfitting and leakage.

## 11. Risks of Inappropriate Encoding

### Risk 1: False Ordering

Using label encoding for nominal categories can introduce an artificial order.

Example: IT = 2 and HR = 1 does not mean IT is twice HR.

### Risk 2: Excessive Features

One-hot encoding a high-cardinality column can create thousands of features, increasing memory requirements and computational cost.

### Risk 3: Target Leakage

Target encoding using information from validation or test targets allows the model to access information it should not have during training.

### Risk 4: Unknown Categories

A production dataset may contain categories not seen during training. Without a defined strategy, the encoder may fail or produce inconsistent features.

### Risk 5: Overfitting

Target encoding can assign extreme target values to categories with only one or two observations.

### Risk 6: Inconsistent Mappings

If categories are mapped differently during training and production, the model may interpret the values incorrectly.

### Risk 7: Losing Meaning

Using arbitrary numerical values can hide important differences between nominal, ordinal, and binary variables.

## 12. Choosing the Right Encoding Technique

| Variable type or situation                      | Suitable starting technique                               | Reason                                      |
| ----------------------------------------------- | --------------------------------------------------------- | ------------------------------------------- |
| Nominal, few categories                         | One-hot encoding                                          | Avoids artificial ordering                  |
| Nominal, linear model with a reference category | Dummy encoding                                            | Supports baseline comparisons               |
| Ordinal with known order                        | Ordinal encoding                                          | Preserves the category order                |
| Binary variable                                 | Explicit 0/1 mapping                                      | Simple and interpretable                    |
| Classification target labels                    | Label encoding                                            | Provides numeric class labels when required |
| High-cardinality category                       | Frequency encoding or carefully validated target encoding | Controls feature dimensionality             |
| Category with unknown future values             | Encoder with an explicit unknown-category policy          | Improves robustness                         |
| Identifier with many unique values              | Investigate usefulness before encoding                    | Avoids unnecessary features and overfitting |

The best encoding technique depends on the feature's meaning, number of categories, algorithm, and deployment requirements.

## 13. Encoding in a Machine Learning Workflow

A typical workflow is:

**Inspect Categories → Identify Variable Type → Choose Encoding → Fit Encoder on Training Data → Transform Training Data → Transform Validation/Test Data → Handle Unknown Categories → Evaluate Model**

The encoder should be fitted only on the training data. Apply the fitted encoder consistently to validation, test, and production data.

## 14. Key Takeaway

Categorical encoding converts categories into numerical representations for Machine Learning.

* Use one-hot encoding for many nominal variables.
* Use ordinal encoding when a meaningful order exists.
* Use explicit binary mappings for two-category variables.
* Consider frequency or target encoding for high-cardinality features.
* Handle unknown categories consistently.
* Prevent target leakage and overfitting.

**Remember:** Select an encoding technique based on the meaning of the variable, not merely because the data contains text.


# Demonstration of the Concept – Categorical Encoding

## Real-World Example: Customer Purchase Prediction

Imagine an e-commerce company wants to predict whether a customer will purchase a product.

The dataset contains:

* Customer ID
* Department
* Education Level
* Membership Status
* Payment Method
* Purchase Result

The dataset contains nominal, ordinal, and binary variables. Each type requires an appropriate encoding strategy.

### Sample Dataset

| Customer ID | Department | Education  | Membership | Payment Method | Purchased |
| ----------- | ---------- | ---------- | ---------- | -------------- | --------: |
| 101         | IT         | Bachelor's | Yes        | UPI            |         1 |
| 102         | HR         | Master's   | No         | Card           |         0 |
| 103         | Finance    | PhD        | Yes        | Cash           |         1 |
| 104         | IT         | School     | No         | UPI            |         0 |
| 105         | HR         | Bachelor's | Yes        | Card           |         1 |

## 1. Identify the Variable Types

### Department – Nominal

Department values include IT, HR, and Finance.

These categories do not have a natural order.

**Technique:** One-hot encoding.

### Education – Ordinal

Education values include School, Bachelor's, Master's, and PhD.

They have a meaningful order.

**Technique:** Ordinal encoding with an explicitly defined order.

### Membership – Binary

Membership contains Yes and No.

**Technique:** Map Yes to 1 and No to 0.

### Payment Method – Nominal

Payment methods include UPI, Card, and Cash.

They have no natural ranking.

**Technique:** One-hot encoding.

### Purchased – Target Variable

The target contains 0 and 1.

* 0 means the customer did not purchase.
* 1 means the customer purchased.

The target is already represented numerically, so it does not need additional categorical encoding.

## 2. Demonstrate One-Hot Encoding

Suppose the Department column contains IT, HR, and Finance.

One-hot encoding creates separate columns for these categories.

A customer from IT receives 1 in `Department_IT` and 0 in the other department columns.

This allows a model to use department membership without assigning an artificial ranking to the departments.

## 3. Demonstrate Ordinal Encoding

Education has a known order:

School < Bachelor's < Master's < PhD

We can map these categories to 0, 1, 2, and 3.

This preserves the order, but the numerical gaps should not automatically be interpreted as equal educational differences.

## 4. Demonstrate Binary Encoding

Membership can be represented as:

* No → 0
* Yes → 1

The mapping is simple and consistent.

## 5. Demonstrate Frequency Encoding

Suppose IT appears 10 times, HR appears 5 times, and Finance appears 2 times.

Frequency encoding replaces the department with its occurrence count:

* IT → 10
* HR → 5
* Finance → 2

This reduces the number of features, but categories with the same frequency may become indistinguishable.

## 6. Demonstrate Target Encoding

Suppose the purchase rates in the training data are:

* IT → 0.60
* HR → 0.40
* Finance → 0.20

These values represent the observed purchase rates for each department in the training data.

Target encoding can be useful for high-cardinality features, but it must be implemented carefully.

If the purchase rate is calculated using validation or test labels, target leakage occurs. Rare categories can also receive unreliable estimates.

## 7. Handling Unknown Categories

Suppose the training data contains IT, HR, and Finance, but new production data contains Marketing.

The encoder needs a defined policy for this unseen category.

For one-hot encoding, an encoder configured with `handle_unknown="ignore"` can transform the unseen category without raising an error. It represents that category with zeros across the known category columns.

This prevents a transformation error, but it does not teach the model what Marketing means. The organization should decide whether the category is valid, should be grouped, or needs separate investigation.

## 8. Final Learning

| Technique          | Example                                | When to use                                     |
| ------------------ | -------------------------------------- | ----------------------------------------------- |
| Label encoding     | Class labels → integer codes           | Target labels or suitable algorithms            |
| Ordinal encoding   | Education → ordered numbers            | Categories with meaningful order                |
| One-hot encoding   | Department → binary columns            | Nominal features with manageable cardinality    |
| Dummy encoding     | Categories with one reference category | Suitable regression and statistical models      |
| Binary encoding    | Yes/No → 1/0                           | Two-category variables                          |
| Frequency encoding | Department → occurrence frequency      | Some high-cardinality features                  |
| Target encoding    | Department → training purchase rate    | High-cardinality features with leakage controls |

**Important:** These techniques are not interchangeable. Always consider the variable type, model requirements, and risk of information leakage before choosing an encoding method.


```python
import pandas as pd

from sklearn.preprocessing import (
    LabelEncoder,
    OrdinalEncoder,
    OneHotEncoder
)

# Step 1: Create a sample customer dataset
df = pd.DataFrame({
    "Customer_ID": [101, 102, 103, 104, 105, 106],
    "Department": [
        "IT", "HR", "Finance", "IT", "HR", "Finance"
    ],
    "Education": [
        "Bachelor's", "Master's", "PhD",
        "School", "Bachelor's", "Master's"
    ],
    "Membership": [
        "Yes", "No", "Yes", "No", "Yes", "No"
    ],
    "Payment_Method": [
        "UPI", "Card", "Cash", "UPI", "Card", "Cash"
    ],
    "Purchased": [1, 0, 1, 0, 1, 0]
})

print("Original Dataset:")
print(df)

# Step 2: Binary Encoding
df["Membership_Encoded"] = df["Membership"].map({
    "No": 0,
    "Yes": 1
})

print("\nBinary Encoding:")
print(df[["Membership", "Membership_Encoded"]])

# Step 3: Ordinal Encoding
education_order = [
    "School",
    "Bachelor's",
    "Master's",
    "PhD"
]

ordinal_encoder = OrdinalEncoder(
    categories=[education_order]
)

df["Education_Encoded"] = ordinal_encoder.fit_transform(
    df[["Education"]]
).astype(int)

print("\nOrdinal Encoding:")
print(df[["Education", "Education_Encoded"]])

# Step 4: One-Hot Encoding for nominal variables
# Fit the encoder on this example's feature data.
# In a real ML workflow, fit it on training data only.
one_hot_encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)

nominal_columns = ["Department", "Payment_Method"]

encoded_array = one_hot_encoder.fit_transform(
    df[nominal_columns]
)

encoded_columns = one_hot_encoder.get_feature_names_out(
    nominal_columns
)

encoded_df = pd.DataFrame(
    encoded_array,
    columns=encoded_columns,
    index=df.index
)

print("\nOne-Hot Encoded Data:")
print(encoded_df)

# Step 5: Dummy Encoding with pandas
dummy_df = pd.get_dummies(
    df["Department"],
    prefix="Department",
    drop_first=True,
    dtype=int
)

print("\nDummy Encoding:")
print(dummy_df)

# Step 6: Label Encoding demonstration
# Demonstration only: Department is nominal, so these codes
# should not automatically be treated as meaningful numerical values.
label_encoder = LabelEncoder()

df["Department_Label"] = label_encoder.fit_transform(
    df["Department"]
)

print("\nLabel Encoding Demonstration:")
print(df[["Department", "Department_Label"]])

# Step 7: Frequency Encoding
department_frequency = df["Department"].value_counts()

df["Department_Frequency"] = (
    df["Department"].map(department_frequency)
)

print("\nFrequency Encoding:")
print(
    df[
        [
            "Department",
            "Department_Frequency"
        ]
    ]
)

# Step 8: Demonstrate simple target encoding
# This example uses training-data category means for illustration.
# For model training, use cross-fitting/out-of-fold target encoding
# to avoid target leakage and overfitting.
department_purchase_rate = df.groupby(
    "Department"
)["Purchased"].mean()

df["Department_Target_Encoded"] = (
    df["Department"].map(department_purchase_rate)
)

print("\nTarget Encoding Demonstration:")
print(
    df[
        [
            "Department",
            "Purchased",
            "Department_Target_Encoded"
        ]
    ]
)

# Step 9: Handle an unknown category
new_customers = pd.DataFrame({
    "Department": ["Marketing"],
    "Payment_Method": ["Wallet"]
})

unknown_encoded = one_hot_encoder.transform(new_customers)

unknown_encoded_df = pd.DataFrame(
    unknown_encoded,
    columns=encoded_columns
)

print("\nUnknown Category Example:")
print(new_customers)

print("\nEncoded Unknown Categories:")
print(unknown_encoded_df)

# Step 10: Display the final dataset
print("\nFinal Dataset with Encoding Demonstrations:")
print(df)
```
